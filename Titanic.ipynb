```python
plt.figure(figsize=(8, 6))
sns.histplot(df['Age'], kde=True, bins=40)
plt.title('Age disribution of Titanic Passenger')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
# we should use median to fill the missing value
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='Embarked')
plt.title('Embarked disribution of Titanic Passenger')
plt.xlabel('Embarked')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
# we should use mode to fill the missing value
```

```python
age_median = df['Age'].median()
df['Age'] = df['Age'].fillna(age_median)


embarked_mode = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(embarked_mode)

df = df.drop(columns=['Cabin'])

df.head(10)
```
